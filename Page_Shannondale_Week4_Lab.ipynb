`break` leaves the loop immediately and `continue` skips the rest of the current pass. Both are tested inside a loop with an if statement.


**7a.** Write a while loop over the numbers 0 through 9 that skips the value 5 using `continue` and stops entirely at the value 8 using `break`. Print the values that survive, separated by spaces.

**Expected output**

```
0 1 2 3 4 6 7
```


````python
n = 0

# TODO: while n < 10, skip 5 with continue and stop at 8 with break
while n < 10:
    if n == 5:
        n += 1
        continue
    if n == 8:
        break

# TODO: print the surviving values on one line using print(n, end=' ')
    print(n, end=' ')
    n += 1
````

**7b.** Review the transactions in order. Skip any transaction of 0 with `continue`, and when a transaction above 200 is found, print `Flagged for review: 210` and stop the loop with `break`. Print each transaction that is checked and passes.

**Expected output**

```
Checked: 45
Checked: 80
Checked: 30
Flagged for review: 210
```


````python
transactions = [45, 0, 80, 30, 210, 60]

# TODO: for loop with continue for zero values and break for values above 200
for transaction in transactions:
  if transaction == 0:
    continue
  if transaction > 200:
    print(f"Flagged for review: {transaction}")
    break
  print(f"Checked: {transaction}")
````

**Expected output**


````python
# TODO: def average_sales(values) returning the rounded average
def classify_order(total):
  if total >= 200:
    return "is a large order"
  elif total >= 75:
    return "is a medium order"
  else:
    return "is a small order"
# TODO: for loop printing each total and its classification
def average_sales(values):
  if not values:
    return 0.0
  return round(sum(values) / len(values), 2)

order_totals = [320, 90, 40, 150, 210]

for total in order_totals:
  classification = classify_order(total)
  print(f"{total} -> {classification}")

# TODO: print the average
avg = average_sales(order_totals)
print(f"Average order total: {avg}")
````
