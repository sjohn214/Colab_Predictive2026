## Question 1 — Creating and accessing a Series

A Series is a one-dimensional labeled array. It has two parts: the **data** (the values)
and the **index** (a label for each value).

**(a)** Build a Series called `weekly_sales` from the `values` list below, using `days` as
the index and `'sales'` as the name.

**(b)** Build a Series called `prices` from the dictionary provided. Notice you don't pass
an index — pandas takes it from the dictionary keys.

**(c)** Print the `.values`, `.index`, `.name`, and `.dtype` of `weekly_sales`.

**(d)** Using **position**, print the first value, then the first three values.

**(e)** Using **labels**, print the value for `'Wed'`, then the values for `'Mon'` and
`'Fri'` together, then everything from `'Tue'` through `'Thu'`.

```python
import numpy as np
import pandas as pd

# (a)
values = [120, 95, 143, 87, 160]
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
prices_dict = {'Mon': 10.5, 'Tue': 11.0, 'Wed': 10.75, 'Thu': 11.25, 'Fri': 12.0}

weekly_sales = pd.Series(data=np.array(values, dtype=np.int32), index=days, name='sales')
print(weekly_sales)
print()
```

```python
# (b)
price_dict = {'pen': 1.50, 'notebook': 3.99, 'stapler': 5.25}

prices = pd.Series(prices_dict)
print(prices)
print()
```

```python
# (c) the parts a Series is made of
print('values:', [weekly_sales.values])
print('index :', [weekly_sales.index.to_numpy()])
print('name  :', [weekly_sales.name])
print('dtype :', [weekly_sales.dtype])
print()
```

```python
# (d) access by POSITION
print(f"first value: {weekly_sales.iloc[0]}")   # the first value only
print(f"\nfirst three values")   # the first three values
print(weekly_sales.iloc[:3])
print()
```

```python
# (e) access by LABEL
print(f"value for 'Wed': {weekly_sales.loc['Wed']}")   # the value for 'Wed'
print("\nvalues for 'Mon' and 'Fri':")   # the values for 'Mon' and 'Fri'
print(weekly_sales.loc[['Mon', 'Fri']])
print("\nvales from 'Tue' through 'Thu':")   # everything from 'Tue' through 'Thu'
print(weekly_sales.loc['Tue':'Thu'])
print()
```

## Question 7 — Boolean (conditional) selection

A comparison on a column produces a Series of `True`/`False`, and you can use that to pick
rows. Use the original `orders` table for this question.

**(a)** Print the boolean mask for `qty >= 5` on its own, so you can see what pandas is
actually working with. Then use it to select those rows.

**(b)** Select rows where `price` is above 3 **and** `qty` is at least 3.

**(c)** Select rows where the category is `bags` **or** `electronics`.

**(d)** Do the same as (c), but with `.isin()` instead of `|`.

**(e)** Select rows where the item is a pen or a notebook **and** the quantity is more than
4, returning only the `order_id`, `item`, and `qty` columns.

```python
import pandas as pd
import numpy as np

orders = pd.DataFrame({
    'order_id': [1001, 1002, 1004, 1007, 1008, 1010, 1013, 1015],
    'item':     ['notebook', 'pen', 'stapler', 'notebook',
                 'backpack', 'pen', 'calculator', 'backpack'],
    'category': ['paper', 'writing', 'office', 'paper',
                 'bags', 'writing', 'electronics', 'bags'],
    'qty':      [3, 12, 2, 5, 1, 8, 2, 3],
    'price':    [3.99, 1.50, 5.25, 3.99, 24.99, 1.50, 12.75, 24.99]
})
# (a) the mask itself, then the rows it selects
# Later parts: give each mask its own name (mask_b, mask_c, ...) so that an old
# mask can never linger and hand you the previous part's answer.
mask_a = orders['qty'] >= 5
print("(a) Mask:\n", mask_a)
print("\n(a) selected rows:\n", orders[mask_a], "\n" + "="*40)
print()
```

```python
# (b) two conditions with AND -- wrap each condition in its own parentheses
mask_b = (orders['price'] > 3) & (orders['qty'] >= 3)
print("(b):\n", orders[mask_b], "\n" + "="*40)
print()
```

```python
# (c) two conditions with OR
mask_c = (orders['category'] == 'bags') | (orders['category'] == 'electronics')
print("(c):\n", orders[mask_c], "\n" + "="*40)
print()
```

```python
# (d) the same result using .isin()
mask_d = orders['category'].isin(['bags', 'electronics'])
print("(d):\n", orders[mask_d], "\n" + "="*40)
print()
```

```python
# (e) combine .isin() with a numeric condition, and pick columns too
mask_e = orders['item'].isin(['pen', 'notebook']) & (orders['qty'] > 4)
columns_e = ['order_id', 'item', 'qty']
print("(e):\n", orders.loc[mask_e, columns_e])
print()
```

```python
returns = pd.DataFrame({
    'days_to_return': [5.0, np.nan, 12.0, np.nan, 8.0],
    'reason': ['damaged', 'wrong item', None, None, 'too small'],
    'refund': [11.97, 18.00, np.nan, np.nan, 24.99]
})
returns
```

```python
import pandas as pd
import numpy as np

returns = pd.DataFrame({
    'days_to_return': [5.0, np.nan, 12.0, np.nan, 8.0],
    'reason': ['damaged', 'wrong item', None, None, 'too small'],
    'refund': [11.97, 18.00, np.nan, np.nan, 24.99]
})

# (a) where are the gaps?
print(returns.isna)           # the full True/False grid
print(returns.isna().sum())           # count per column
print('total missing:', returns.isna().sum().sum())
```

```python
# (b) drop rows only if the entire row is missing
clean_returns =returns.dropna(how='all')
print(clean_returns)
print()
```

```python
# (c) drop rows missing a value in specific columns
clean_subset = returns.dropna(subset=['days_to_return', 'refund'], how='any')
print(clean_subset)
print()
```

```python
# (d) drop columns containing any missing value
clean_columns = returns.dropna(axis=1)
print(clean_columns)
print()
```

**(d) Your answer:** What did you get back, and why did that happen with this particular
table? When would dropping columns be a reasonable choice?

*Every column contains at least one missing value and because pandas defaults by dropping columns missing any data, all three were removed. This left an empty structure with just the original row index behind.*

```python
import pandas as pd
import numpy as np

returns = pd.DataFrame({
    'days_to_return': [5.0, np.nan, 12.0, np.nan, 8.0],
    'reason': ['damaged', 'wrong item', None, None, 'too small'],
    'refund': [11.97, 18.00, np.nan, np.nan, 24.99]
})


# (e) three different filling strategies
filled = returns.copy()
filled['days_to_return'] = filled['days_to_return'].fillna(filled['days_to_return'].mean())
filled['reason'] = filled['reason'].ffill()
filled['refund'] = filled['refund'].bfill()
print(filled)
print()
```

```python
# use None into the reason column but np.nan into the numeric columns.
import pandas as pd
import numpy as np

# Recreating the table using explicit None and np.nan types
returns = pd.DataFrame({
    'days_to_return': [5.0, np.nan, 12.0, np.nan, 8.0],
    'reason': ['damaged', 'wrong item', None, None, 'too small'],
    'refund': [11.97, 18.00, np.nan, np.nan, 24.99]
})

print(returns.isna())
print()
```
